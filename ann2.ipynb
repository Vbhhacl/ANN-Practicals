```python
import numpy as np

def mcculloch_pitts_neuron(inputs, weights, threshold):
    """
    McCulloch-Pitts neuron model with a step activation function.
    """
    # Calculate the weighted sum of inputs
    weighted_sum = np.dot(inputs, weights) #

    # Apply the threshold activation function
    if weighted_sum >= threshold: #
        return 1
    else:
        return 0

def andnot_function():
    """
    Implements the ANDNOT function using a McCulloch-Pitts neuron with user input.
    """
    # Define optimal weights and threshold for the ANDNOT function
    # Weights: w1=1 (excitatory), w2=-1 (inhibitory)
    # Threshold: theta=1
    weights = np.array([1, -1])
    threshold = 1

    print("--- McCulloch-Pitts Neuron for ANDNOT Function ---")
    print("Truth Table:")
    print("Input 1 | Input 2 | Output")
    print("-" * 25)

    # Test with all possible binary input combinations
    test_inputs = np.array([
        [0, 0],
        [0, 1],
        [1, 0],
        [1, 1]
    ])

    for inputs in test_inputs:
        output = mcculloch_pitts_neuron(inputs, weights, threshold)
        print(f"   {inputs[0]}    |    {inputs[1]}    |   {output}")

    # Take user input
    while True:
        try:
            user_input1 = int(input("\nEnter first binary input (0 or 1): "))
            user_input2 = int(input("Enter second binary input (0 or 1): "))
            if user_input1 in [0, 1] and user_input2 in [0, 1]:
                user_inputs = np.array([user_input1, user_input2])
                user_output = mcculloch_pitts_neuron(user_inputs, weights, threshold)
                print(f"\nFor inputs {user_inputs}, the ANDNOT output is: {user_output}")
                break
            else:
                print("Invalid input. Please enter only 0 or 1.")
        except ValueError:
            print("Invalid input. Please enter numerical values.")

if __name__ == "__main__":
    andnot_function()
```
