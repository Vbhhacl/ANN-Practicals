```python
import numpy as np
import matplotlib.pyplot as plt

# Create input values
x = np.linspace(-10, 10, 1000)

# Activation Functions
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def tanh(x):
    return np.tanh(x)

def relu(x):
    return np.maximum(0, x)

def leaky_relu(x, alpha=0.01):
    return np.where(x > 0, x, alpha * x)

def step_function(x):
    return np.where(x >= 0, 1, 0)

# Plotting
plt.figure(figsize=(12, 8))

plt.plot(x, sigmoid(x), label="Sigmoid")
plt.plot(x, tanh(x), label="Tanh")
plt.plot(x, relu(x), label="ReLU")
plt.plot(x, leaky_relu(x), label="Leaky ReLU")
plt.plot(x, step_function(x), label="Step Function")

plt.title("Common Activation Functions in Neural Networks")
plt.xlabel("Input")
plt.ylabel("Output")
plt.legend()
plt.grid(True)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 1000)

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def tanh(x):
    return np.tanh(x)

def relu(x):
    return np.maximum(0, x)

def softmax(x):
    # Softmax of vector [x, 0]
    e_x = np.exp(np.vstack([x, np.zeros_like(x)]))
    return e_x[0] / e_x.sum(axis=0)

def step_function(x):
    return np.where(x >= 0, 1, 0)

plt.figure(figsize=(12, 8))

plt.plot(x, sigmoid(x), linestyle='-',  color='black', linewidth=2, label="Sigmoid")
plt.plot(x, tanh(x), linestyle='--', color='black', linewidth=2, label="Tanh")
plt.plot(x, relu(x), linestyle='-.', color='black', linewidth=2, label="ReLU")
plt.plot(x, softmax(x), linestyle=':', color='black', linewidth=2, label="Softmax")
plt.plot(x, step_function(x), linestyle=(0, (5, 1)), color='black', linewidth=2, label="Step Function")

plt.legend(loc='upper left', fontsize=12)

# Labels with spacing
plt.text(3.4, sigmoid(3.4) + 0.05, "Sigmoid", fontsize=11)
plt.text(2.2, tanh(2.2) - 0.15, "Tanh", fontsize=11)
plt.text(2.5, relu(2.5) + 0.3, "ReLU", fontsize=11)
plt.text(1.2, softmax(1.2) + 0.02, "Softmax", fontsize=11)
plt.text(0.2, 1.05, "Step Function", fontsize=11)

plt.title("Common Activation Functions in Neural Networks")
plt.xlabel("Input")
plt.ylabel("Output")
plt.ylim(-1.5, 5.5)
plt.grid(False)

plt.show()
```
