```python
import numpy as np

class HopfieldNetwork:
    def __init__(self, num_input_features, num_output_features):
        # Initialize the weight matrix with zeros
        self.weights = np.zeros((num_input_features, num_output_features))

    def sign(self, x):
        # Sign activation function: returns 1 for x >= 0, and -1 for x < 0
        return np.where(x >= 0, 1, -1)

    def train(self, X_patterns, Y_patterns):
        # Train the network using the outer product rule
        # X_patterns: input patterns, Y_patterns: corresponding output patterns
        for i in range(len(X_patterns)):
            # For each pattern pair, update weights using outer product
            self.weights += np.outer(X_patterns[i], Y_patterns[i])
        print("\nWeight Matrix W:")
        print(self.weights)

    def forward_recall(self, x_test):
        # Perform forward recall: given an input pattern, retrieve the associated output pattern
        # x_test: the input pattern to test
        y_out = self.sign(np.dot(x_test, self.weights))
        print("\nForward Recall")
        print("Input X:", x_test)
        print("Output Y:", y_out)
        return y_out

    def backward_recall(self, y_retrieved):
        # Perform backward recall: given an output pattern, retrieve the associated input pattern
        # y_retrieved: the output pattern to use for recall
        # W.T is used for backward recall (transpose of the weight matrix)
        x_recall = self.sign(np.dot(y_retrieved, self.weights.T))
        print("\nBackward Recall")
        print("Recovered X:", x_recall)
        return x_recall
print("Name:Vaibhavi Halloli")
print("Role no. TE19")
# Define training patterns
X = np.array([
    [1, 1, -1],
    [-1, 1, 1]
])

Y = np.array([
    [1, 1],
    [-1, 1]
])

# Create a Hopfield Network instance
hopfield_net = HopfieldNetwork(X.shape[1], Y.shape[1])

# Train the network
hopfield_net.train(X, Y)

# Test Forward Recall with a known pattern
x_test_pattern = np.array([1, -1, 1])
y_output_pattern = hopfield_net.forward_recall(x_test_pattern)

# Test Backward Recall with the retrieved output pattern
hopfield_net.backward_recall(y_output_pattern)
```
