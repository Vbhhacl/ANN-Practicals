```python
# Perceptron Program to recognize Even and Odd numbers
# ASCII values of digits 0-9 are from 48 to 57

import random

# Step 1: Training Data
# Input = ASCII value
# Output = 0 (Even), 1 (Odd)

training_data = [
    (48, 0),  # '0'
    (49, 1),  # '1'
    (50, 0),  # '2'
    (51, 1),  # '3'
    (52, 0),  # '4'
    (53, 1),  # '5'
    (54, 0),  # '6'
    (55, 1),  # '7'
    (56, 0),  # '8'
    (57, 1)   # '9'
]

# Step 2: Initialize weight and bias randomly
weight = random.uniform(-1, 1)
bias = random.uniform(-1, 1)

learning_rate = 0.01

# Step 3: Activation Function
def step_function(x):
    if x >= 0:
        return 1
    else:
        return 0

# Step 4: Training the perceptron
for epoch in range(100):
    for ascii_value, target in training_data:

        # Calculate output
        linear_output = weight * ascii_value + bias
        prediction = step_function(linear_output)

        # Calculate error
        error = target - prediction

        # Update weight and bias
        weight = weight + learning_rate * error * ascii_value
        bias = bias + learning_rate * error

# Step 5: Testing
print("Training Completed!")
print("Final Weight:", weight)
print("Final Bias:", bias)

# Take user input
num = input("Enter a digit (0-9): ")

ascii_val = ord(num)  # Convert to ASCII

result = step_function(weight * ascii_val + bias)

if result == 0:
    print("Number is EVEN")
else:
    print("Number is ODD")
```
